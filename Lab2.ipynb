Реализовать следующие структуры данных:
  
● <b> Стек (stack): </b>  
Операции для стека:
<i> инициализация, проверка на пустоту, добавление нового элемента в начало, извлечение элемента из начала;  </i>  
  
● <b> Дек (двусторонняя очередь, deque):  </b>  
Операции для дека:
<i> инициализация, проверка на пустоту, добавление нового элемента в начало, добавление нового элемента в конец, извлечение элемента из начала, извлечение элемента из конца. </i>  
  
<i> Разработать программу обработки данных, содержащихся в заранее подготовленном txt-файле, в соответствии с заданиями, применив указанную в задании структуру данных. Результат работы программы вывести на экран и сохранить в отдельном txt-файле. </i>

```python
# класс стек - последним пришёл – первым вышел. 
class MyStack:
    def __init__(self, name="Stack"):
        self.items = []
        self.name = name

    def is_empty(self):
        return len(self.items) == 0

    # добавление в начало
    def push(self, item):
        self.items.insert(0, item)

    # извлечение из начала
    def pop(self):
        if not self.is_empty():
            return self.items.pop(0)
        else:
            return None
```

```python
# класс дек - структура данных, в которой добавлять и удалять элементы можно с обоих концов.
class MyDeque:
    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    # добавление в начало
    def push_front(self, item):
        self.items.insert(0, item)

    # добавление в конец
    def push_back(self, item):
        self.items.append(item)

    # Извлечение из начала
    def pop_front(self):
        if not self.is_empty():
            return self.items.pop(0)
        else:
            return None

    # Извлечение из конца
    def pop_back(self):
        if not self.is_empty():
            return self.items.pop()
        else:
            return None
```

### Задание №1
Отсортировать строки файла, содержащие названия книг, в алфавитном порядке с использованием двух деков.

```python
def sort_books():
    # чтение названий книг из файла
    with open('books.txt', 'r', encoding='utf-8') as f:
        lines = [line.strip() for line in f if line.strip() != ""]
    
    # экз дека, для неотсортированных книг
    unsorted_deque = MyDeque()
    for book in lines:
        unsorted_deque.push_back(book)
    
    # экз дека, для отсортированных книг
    sorted_deque = MyDeque()
    
    # функция для вставки строки в отсортированный дек
    def sorted_insert(deq, book):
        temp = MyDeque()
        # перемещаем из отсортированного дека те элементы, которые идут до места вставки
        while (not deq.is_empty()) and (deq.items[0] < book):
            temp.push_back(deq.pop_front())
        deq.push_front(book)
        # возвращаем временные элементы обратно в отсортированный дек
        while not temp.is_empty():
            deq.push_front(temp.pop_back())
    
    # извлекаем из неотсортированного дека и вставляем в отсортированный
    while not unsorted_deque.is_empty():
        book = unsorted_deque.pop_front()
        sorted_insert(sorted_deque, book)
    
    # вывод результата
    sorted_books = sorted_deque.items
    print("Отсортированные книги:")
    for book in sorted_books:
        print(book)
    
    with open('books_sorted.txt', 'w', encoding='utf-8') as f:
        for book in sorted_books:
            f.write(book + "\n")

# вызов:
sort_books()
```

### Задание №2
Дек содержит последовательность символов для шифровки сообщений. Дан текстовый файл, содержащий зашифрованное сообщение. Пользуясь деком, расшифровать текст.
Известно, что при шифровке каждый символ сообщения заменялся следующим за ним в деке по часовой стрелке через один.

```python
def decrypt_message():
    # чтение строки-шифра из файла
    with open('cipher.txt', 'r', encoding='utf-8') as f:
        cipher_str = f.read().strip()
    
    # заполнение дека символами шифра
    cipher_deque = MyDeque()
    for ch in cipher_str:
        cipher_deque.push_back(ch)
    
    # чтение зашифрованного сообщения
    with open('encrypted.txt', 'r', encoding='utf-8') as f:
        encrypted_message = f.read().strip()
    
    # пустая строка для хранения расшифровки 
    decrypted_message = ""
    for ch in encrypted_message:
        try:
            idx = cipher_deque.items.index(ch)
        except ValueError:
            # Если символ не найден в шифре, оставляем его без изменений
            decrypted_message += ch
            continue
        # Вычисляем предыдущий индекс с цикличностью
        prev_idx = (idx - 1) % len(cipher_deque.items)
        decrypted_message += cipher_deque.items[prev_idx]
    
    print("Расшифрованное сообщение:")
    print(decrypted_message)
    
    with open('decrypted.txt', 'w', encoding='utf-8') as f:
        f.write(decrypted_message)

# Пример вызова:
decrypt_message()
```

### Задание №3
Даны три стержня и n дисков различного размера. Диски можно надевать на стержни, образуя из них башни. Перенести n дисков со стержня А на стержень С, сохранив их первоначальный порядок. При переносе дисков необходимо соблюдать следующие правила:
- на каждом шаге со стержня на стержень переносить только один диск;
- диск нельзя помещать на диск меньшего размера;
- для промежуточного хранения можно использовать стержень В.
Реализовать алгоритм, используя три стека вместо стержней А, В, С. Информация о дисках хранится в исходном файле.

```python
def towers_of_hanoi(n, source, auxiliary, destination, moves):
    if n == 1: # если остался только один диск, с исходного стержня на целевой
        disk = source.pop()
        destination.push(disk)
        moves.append(f"Переместить диск {disk} со стержня {source.name} на стержень {destination.name}")
    else:
        towers_of_hanoi(n-1, source, destination, auxiliary, moves) # перемещаем n-1 дисков с исходного стержня на вспомогательный исп цел стрежень как вспомог
        disk = source.pop() # извлекаем диск с вершины исходного стержня
        destination.push(disk) # перемещаем его на целевой стержень
        moves.append(f"Переместить диск {disk} со стержня {source.name} на стержень {destination.name}")
        towers_of_hanoi(n-1, auxiliary, source, destination, moves) # перемещаем n-1 дисков со вспом стержня на цел

def solve_hanoi():
    # Чтение информации о дисках из файла
    with open('disks.txt', 'r', encoding='utf-8') as f:
        disks = f.read().split()
    disks = list(map(int, disks))
    # Переворачиваем список, чтобы на вершине был наименьший диск
    disks = disks[::-1]
    
    # три стержня A, B и C в виде стеков
    A = MyStack("A")
    B = MyStack("B")
    C = MyStack("C")
    for d in disks:
        # Добавляем диски в стек A.
        A.push(d)
    
    # список ходов
    moves = []
    towers_of_hanoi(len(disks), A, B, C, moves)
    
    print("Ходы решения задачи о Ханойских башнях:")
    for move in moves:
        print(move)
    
    with open('hanoi_moves.txt', 'w', encoding='utf-8') as f:
        f.write("\n".join(moves))

# Пример вызова:
solve_hanoi()
```

### Задание №4
Дан текстовый файл с программой на алгоритмическом языке. За один просмотр файла проверить баланс круглых скобок в тексте, используя стек.

```python
def check_round_brackets(filename):
    stack = MyStack("BracketStack") # для хранения открывающихся скобок
    with open(filename, 'r', encoding='utf-8') as f:
        content = f.read()
    
    # изначально balanced
    balanced = True
    for ch in content:
        if ch == '(': 
            stack.push(ch)
        elif ch == ')':
            if stack.is_empty():
                balanced = False # если стек пуст, значит нет ( - no balanced 
                break
            else:
                stack.pop() # если стек не пуст, значит есть ( - balanced 
    
    # если остались открытые ( то не нашлось ) - no balanced
    if not stack.is_empty():
        balanced = False
    
    if balanced:
        print("Круглые скобки сбалансированы.")
    else:
        print("Круглые скобки не сбалансированы.")

# Пример вызова:
check_round_brackets("program_round.txt")
```

### Задание №5
Дан текстовый файл с программой на алгоритмическом языке. За один просмотр файла проверить баланс квадратных скобок в тексте, используя дек.

```python
def check_square_brackets(filename):
    deque_ = MyDeque()
    with open(filename, 'r', encoding='utf-8') as f:
        content = f.read()
    
    balanced = True
    for ch in content:
        if ch == '[':
            deque_.push_back(ch)
        elif ch == ']':
            if deque_.is_empty():
                balanced = False
                break
            else:
                deque_.pop_back()
    
    if not deque_.is_empty():
        balanced = False
    
    if balanced:
        print("Квадратные скобки сбалансированы.")
    else:
        print("Квадратные скобки не сбалансированы.")

# Пример вызова:
check_square_brackets("program_square.txt")
```

### Задание №6
Дан файл из символов. Используя стек, за один просмотр файла напечатать сначала все цифры, затем все буквы, и, наконец, все остальные символы, сохраняя исходный порядок в каждой группе символов.

```python
def process_characters(filename):
    digits_stack = MyStack("Digits")
    letters_stack = MyStack("Letters")
    others_stack = MyStack("Others")
    
    with open(filename, 'r', encoding='utf-8') as f:
        content = f.read()
    
    for ch in content:
        if ch.isdigit():
            digits_stack.push(ch) # если цифра - Digits
        elif ch.isalpha():
            letters_stack.push(ch) # если буква - Letters
        else:
            others_stack.push(ch) # ни то и ни то - Others
    
    # Восстанавливаем исходный порядок – поскольку при push порядок инвертируется
    digits = digits_stack.items[::-1]
    letters = letters_stack.items[::-1]
    others = others_stack.items[::-1]
    
    result = "".join(digits + letters + others)
    print("Результат обработки символов:")
    print(result)
    
    with open('processed_chars.txt', 'w', encoding='utf-8') as f:
        f.write(result)

# Пример вызова:
process_characters("symbols.txt")
```

### Задание №7
Дан файл из целых чисел. Используя дек, за один просмотр файла напечатать сначала все отрицательные числа, затем все положительные числа, сохраняя исходный порядок в каждой группе.

```python
def process_numbers(filename):
    negatives = MyDeque()
    positives = MyDeque()
    
    with open(filename, 'r', encoding='utf-8') as f:
        numbers = f.read().split()
    
    numbers = list(map(int, numbers)) # из списка строк в список целых чисел
    
    for num in numbers:
        if num < 0:      
            negatives.push_back(num) # negatives
        elif num > 0:
            positives.push_back(num) # positives
    
    result = negatives.items + positives.items
    result_str = " ".join(map(str, result))
    
    print("Обработанные числа:")
    print(result_str)
    
    with open('processed_numbers.txt', 'w', encoding='utf-8') as f:
        f.write(result_str)

# Пример вызова:
process_numbers("numbers.txt")
```

### Задание №8
Дан текстовый файл. Используя стек, сформировать новый текстовый файл, содержащий строки исходного файла, записанные в обратном порядке: первая строка становится последней, вторая – предпоследней и т.д.

```python
def reverse_file_lines(input_filename, output_filename):
    stack = MyStack("Lines")
    with open(input_filename, 'r', encoding='utf-8') as f:
        lines = f.readlines()
    
    for line in lines:
        # удаляем символ перевода строки с конца строки и помещаем очищенную строку в стек
        stack.push(line.rstrip("\n"))
    
    # список для обратки 
    reversed_lines = []
    while not stack.is_empty(): # пока стек не пуст
        # извлекаем строку с вершины стека и добавляем ее в список
        reversed_lines.append(stack.pop())
    
    with open(output_filename, 'w', encoding='utf-8') as f:
        for line in reversed_lines:
            f.write(line + "\n")
    
    print("Новый файл с перевёрнутыми строками сохранён как", output_filename)

# Пример вызова:
reverse_file_lines("text.txt", "reversed.txt")
```
