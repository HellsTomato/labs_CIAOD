### Задание №2
Написать генератор случайных матриц(многомерных), который принимает
опциональные параметры <b>m</b>, <b>n</b>, <b>min_limit</b>, <b>max_limit</b>, где <b>m</b> и <b>n</b> указывают размер
матрицы, а <b>min_lim</b> и <b>max_lim</b> - минимальное и максимальное значение для

```python
import random
```

```python
user_m = int(input("кол-во строк (m)"))
user_n = int(input("кол-во столбцов (n)"))
user_min_limit = int(input("мин-знач"))
user_max_limit = int(input("макс-знач"))

matrix = []
for i in range(user_m):
    row = [random.randint(user_min_limit, user_max_limit) #случайные числа в диапазоне от min до max.
for _ in range(user_n)]
    matrix.append(row) # строку в матрицу.
for row in matrix:     # Печатаем каждую строку матрицы.
    print(row)
```

### Задание №3
Реализовать методы сортировки строк числовой матрицы в соответствии с
заданием. Оценить время работы каждого алгоритма сортировки и сравнить его со
временем стандартной функции сортировки. Испытания проводить на сгенерированных
матрицах.

```python
import copy
import time
```

```python
# Стандартная сортировка 
start_time = time.time()                 # начальное время вып
matrix_copy = copy.deepcopy(matrix)      # коп матрицы 
for i in range(len(matrix_copy)):        # перебор все строки в матрице где i=инд
    matrix_copy[i] = sorted(matrix_copy[i]) # i = тек строка 
print("--- {0} ms ---".format(round((time.time() - start_time)*1000)))
print(matrix_copy)
```

```python
# Сортировка выбором. o(n^2)
# Берётся срез массива, в котором минимальный элемент переносят в самый левый угол,
# после чего срез уменьшается и цикл повторяется.
start_time = time.time()
matrix_copy = copy.deepcopy(matrix) 

def selection_sort(matrix):
    for row in matrix:                      # прох по каж строк
        for i in range(len(row)):           # перебор всех эл в строке
            min_idx = i                     # i = 0
            for j in range(i+1, len(row)):  # сравниваем тек мин с оставльными 
                if row[j] < row[min_idx]: 
                    min_idx = j 
            row[i], row[min_idx] = row[min_idx], row[i]

selection_sort(matrix_copy)
print("--- {0} ms ---".format(round((time.time() - start_time)*1000)))
print(matrix_copy)
```

```python
# Сортировка вставкой. o(n^2)
# Алгоритм, который сортирует элементы, последовательно вставляя каждый 
# новый элемент в правильное место среди уже отсортированных элементов.
start_time = time.time()

def insertion_sort(matrix):               
    for row in matrix:                     # прох по каж строк
        for i in range(1, len(row)):       # начинаем с первого индекса 
            key = row[i]                   # сохраняем его, чтобы потом вставить
            j = i - 1                      # начинаем поиск места для вставки
            while j >= 0 and key < row[j]: # 1ое ограничитель и 2<9 
                row[j + 1] = row[j]        # сдвигаем 9 с 0 на 1
                j -= 1                     # переход к предыдущиму, если не нашли место
            row[j + 1] = key               # 2,9,7,4

matrix_copy = copy.deepcopy(matrix) 
insertion_sort(matrix_copy)  
print("--- {0} ms ---".format(round((time.time() - start_time)*1000)))
print(matrix_copy)
```

```python
# Сортировка обменом. || Сортировка пузырьком. o(n^2)
start_time = time.time()

def bubble_sort(matrix):
    for row in matrix:             # прох по каж строк
        n = len(row)               # длина строки
        for i in range(n - 1):     # проходим по строке начиная с первого элемента
            for j in range(0, n - i - 1): # сравниваем элементы двигаясь от начала
                if row[j] > row[j + 1]:   # если текущий элемент больше следующего
                    row[j], row[j + 1] = row[j + 1], row[j] # меняем их местами

matrix_copy = copy.deepcopy(matrix) 
bubble_sort(matrix_copy) 
print("--- {0} ms ---".format(round((time.time() - start_time)*1000)))
print(matrix_copy)
```

```python
# Сортировка Шелла - удучшенная сортировка вставками, сначала используя o(n^2)
# большие шаги, а затем постепенно уменьшает шаги для более точной сортировки.
start_time = time.time()

def shell_sort(matrix):
    for row in matrix:                                 # прох по каж строк
        n = len(row)                                  
        gap = n // 2                  # начальный шаг половина длины строки
        while gap > 0:                # пока шаг больше нуля
            for i in range(gap, n):   # идем по всем элементам строки начиная с позиции gap
                temp = row[i]         # сохраняем текущий элемент
                j = i
                while j >= gap and row[j - gap] > temp: # пока не достигли начала строки и текущий элемент меньше соседнего
                    row[j] = row[j - gap] # сдвигаем элемент вправо
                    j -= gap              # переходим на шаг назад
                row[j] = temp             # вставляем элемент на правильное место
            gap //= 2                     # уменьшаем шаг в два раза

matrix_copy = copy.deepcopy(matrix)  
shell_sort(matrix_copy) 
print("--- {0} ms ---".format(round((time.time() - start_time)*1000)))
print(matrix_copy)
```

```python
# Быстрая сортировка - это эффективный алгоритм сортировки, который использует метод o(n*log(n))
# разделяй и властвуй. Он разделяет массив на части вокруг опорного элемента и рекурсивно сортирует эти части. 
start_time = time.time()

def quick_sort(matrix):
    def quick_sort_row(row): # вложенная функция для сортировки строки
        if len(row) <= 1: # если длина равна 1 она уже отсортирована
            return row
        pivot = row[len(row) // 2]                  # выбираем опорный элемент строки
        left = [x for x in row if x < pivot]        # меньше опорного
        middle = [x for x in row if x == pivot]     # равные опорному
        right = [x for x in row if x > pivot]       # больше опорного
        return quick_sort_row(left) + middle + quick_sort_row(right) # рекурсивно сортируем и объединяем

    for i in range(len(matrix)):             # применяем быструю сортировку к каждой строке матрицы
        matrix[i] = quick_sort_row(matrix[i])

matrix_copy = copy.deepcopy(matrix)
quick_sort(matrix_copy)
print("--- {0} ms ---".format(round((time.time() - start_time)*1000)))
print(matrix_copy)
    
```

```python
# Турнирная сортировка - это алгоритм o(n log(n))
# где на каждом шаге выбирается победитель (минимальный элемент) из каждой пары, и таким образом, мы сортируем элементы.
start_time = time.time()

def tournament_sort(matrix):
    def tournament(row):                                   # вложенная функция для сортировки одной строки
        while len(row) > 1:                                # пока в строке больше одного элемента
            new_row = []                                   # новый список для хранения результатов
            for i in range(0, len(row), 2):                # проходим через элементы пары
                if i + 1 < len(row):                       # если есть пара элементов
                    new_row.append(min(row[i], row[i + 1]))# сравниваем пару и добавляем минимальный элемент
                else:
                    new_row.append(row[i])                 # если нечетное количество элементов, добавляем последний
            row = new_row                                  # обновляем строку результатами текущего турнира
        return row[0]                                      # возвращаем единственный оставшийся элемент, который и будет минимальным

    for i in range(len(matrix)):                           # применяем турнирную сортировку ко всем строкам матрицы
        matrix[i] = tournament(matrix[i])
                                                                      
matrix_copy = copy.deepcopy(matrix)                                    # пара 9, 2 — минимальный элемент 2
tournament_sort(matrix_copy)                                           # пара 7, 4 — минимальный элемент 4.
print("--- {0} ms ---".format(round((time.time() - start_time)*1000))) # пара 2, 4 — минимальный элемент 2
print(matrix_copy)
```
