```python
# Генерация данных
def generate_data(size=20, max_value=100):
    arr = [random.randint(0, max_value) for _ in range(size)]
    arr.sort()
    return arr

# Проверка
test_data = generate_data(10, 50)
print("пример:", test_data)
```

```python
# Бинарный поиск - делит отсортированный массив пополам на каждом шаге, что позволяет найти элемент за логарифмическое время O(log n)
def binary_search(arr, x):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == x:
            return mid
        elif arr[mid] < x:
            left = mid + 1
        else:
            right = mid - 1
    return -1

#функция добавления элемента
def add_element_binary(arr, x):
    bisect.insort(arr, x)

#функция удаления элемента
def remove_element_binary(arr, x):
    try:
        arr.remove(x)
    except ValueError:
        pass

# проверка
arr_example = [1, 3, 5, 7, 9]
print("пример:", arr_example)
pos = binary_search(arr_example, 7)
print("проверка:", pos)  # 3
```

```python
# Фибоначчиев поиск - как и бинарный поиск, только место деления списка пополам он использует числа Фибоначчи (0, 1, 1, 2, 3, 5, 8, …).
def fibonacci_search(arr, x):
    n = len(arr)
    fib2 = 0
    fib1 = 1
    fibm = fib2 + fib1
    while fibm < n:
        fib2 = fib1
        fib1 = fibm
        fibm = fib2 + fib1
    offset = -1
    while fibm > 1:
        i = min(offset + fib2, n - 1)
        if arr[i] < x:
            fibm = fib1
            fib1 = fib2
            fib2 = fibm - fib1
            offset = i
        elif arr[i] > x:
            fibm = fib2
            fib1 = fib1 - fib2
            fib2 = fibm - fib1
        else:
            return i
    if fib1 and offset + 1 < n and arr[offset + 1] == x:
        return offset + 1
    return -1

# Короткая проверка
print("пример:", arr_example)
pos_fib = fibonacci_search(arr_example, 9)
print("проверка:", pos_fib)  # 4
```

```python
# Интерполяционный поиск -  рассчитываем предполагаемую позицию искомого элемента, основываясь на том, 
# насколько далеко он может находиться между минимальным и максимальным значениями массива.
def interpolation_search(arr, x):
    low, high = 0, len(arr) - 1
    while low <= high and x >= arr[low] and x <= arr[high]:
        if low == high:
            return low if arr[low] == x else -1
        pos = low + int((float(high - low) / (arr[high] - arr[low])) * (x - arr[low]))
        if arr[pos] == x:
            return pos
        elif arr[pos] < x:
            low = pos + 1
        else:
            high = pos - 1
    return -1

# Короткая проверка
print("пример:", arr_example)
pos_int = interpolation_search(arr_example, 5)
print("проверка:", pos_int)  # 2
```

```python
# Бинарное дерево -  используется для хранения информации в виде узлов, связанных друг с другом. 
# Каждый узел в бинарном дереве имеет до двух потомков: левый и правый. аналогия - работает как хорошо организованный словарь в котором быстро всё можно найти
# левое поддерево знач < текущ узла
# праваое поддерево знач > текущ узла 
class Node:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None

class BinarySearchTree:
    def __init__(self):
        self.root = None

    def insert(self, key):
        if self.root is None:
            self.root = Node(key)
        else:
            self._insert(self.root, key)

    def _insert(self, current_node, key):
        # Если вставляемый ключ меньше - идём в левую ветвь
        if key < current_node.key:
            if current_node.left is None:
                current_node.left = Node(key)
            else:
                self._insert(current_node.left, key)

        # Если вставляемый ключ больше - идём в правую ветвь
        elif key > current_node.key:
            if current_node.right is None:
                current_node.right = Node(key)
            else:
                self._insert(current_node.right, key)
        
        # если ключ равен уже имеющемуся — просто игнорируем (не вставляем дубликат)
        else:
            return  # не вставлять дубль

    def search(self, key):
        return self._search(self.root, key)

    def _search(self, current_node, key):
        if current_node is None:
            return None
        if current_node.key == key:
            return current_node
        elif key < current_node.key:
            return self._search(current_node.left, key)
        else:
            return self._search(current_node.right, key)

    def delete(self, key):
        self.root = self._delete(self.root, key)

    def _delete(self, current_node, key):
        if current_node is None:
            return current_node
        if key < current_node.key:
            current_node.left = self._delete(current_node.left, key)
        elif key > current_node.key:
            current_node.right = self._delete(current_node.right, key)
        else:
            # если мы нашли узел, который нужно удалить

            # 1) Узел не имеет потомков или имеет одного потомка
            if current_node.left is None:
                return current_node.right
            elif current_node.right is None:
                return current_node.left
            
            # 2) Узел имеет двух потомков
            temp = self._min_value_node(current_node.right)
            current_node.key = temp.key
            current_node.right = self._delete(current_node.right, temp.key)
        return current_node

    def _min_value_node(self, node):
        current = node
        while current.left:
            current = current.left
        return current


# Короткая проверка
tree_test = BinarySearchTree()
for v in [8,3,10,1,6,14,4,7,13]:
    tree_test.insert(v)
found_node = tree_test.search(7)
print("найден ли узел 7?", "Да" if found_node else "Нет")
tree_test.delete(7)
found_node_after_delete = tree_test.search(7)
print("После удаления из дерева:", "Найден" if found_node_after_delete else "Не найден")
```

### Задание №2: 
<i>Методы разрешения коллизий в хеш-таблицах</i>

```python
# линейное рехэширование - последовательно проверяет следующие ячейки,  пока не найдёт подходящую ячейку для вставки, поиска или удаления ключа.
class HashTableLinear:
    def __init__(self, size=10):
        self.size = size
        self.table = [None] * size

    def _hash(self, key):
        return key % self.size

    def insert(self, key, value=None):
        idx = self._hash(key)
        start_idx = idx
        while self.table[idx] is not None and self.table[idx][0] != key:
            idx = (idx + 1) % self.size
            if idx == start_idx:
                raise Exception("HashTable is full!")
        self.table[idx] = (key, value)

    def search(self, key):
        idx = self._hash(key)
        start_idx = idx
        while self.table[idx] is not None:
            if self.table[idx][0] == key:
                return self.table[idx]
            idx = (idx + 1) % self.size
            if idx == start_idx:
                return None
        return None

    def delete(self, key):
        idx = self._hash(key)
        start_idx = idx
        while self.table[idx] is not None:
            if self.table[idx][0] == key:
                self.table[idx] = None
                return
            idx = (idx + 1) % self.size
            if idx == start_idx:
                return
```

```python
# псевдослучайное рехэширование - вместо фиксированного шага  использовать последовательность псевдослучайных шагов, определяемых генератором.
# генератор псевдослучайных чисел подсказывает, на сколько позиций сместиться, пока не найдется нужное место или не будет исчерпан весь массив. 
# Это позволяет распределить элементы более равномерно и снизить риск возникновения кластеров в таблице.
class HashTablePseudoRandom:
    def __init__(self, size=10, seed=42):
        self.size = size
        self.table = [None] * size
        self.seed = seed

    def _hash(self, key):
        return key % self.size

    def _step_generator(self):
        # Каждый раз заново инициализируем seed,
        # чтобы генерация псевдослучайных шагов была воспроизводимой.
        random.seed(self.seed)
        while True:
            yield random.randint(1, self.size - 1)

    def insert(self, key, value=None):
        idx = self._hash(key)
        start_idx = idx
        steps = self._step_generator()

        while self.table[idx] is not None and self.table[idx][0] != key:
            step = next(steps)
            idx = (idx + step) % self.size
            if idx == start_idx:
                raise Exception("HashTable is full!")
        self.table[idx] = (key, value)

    def search(self, key):
        idx = self._hash(key)
        start_idx = idx
        steps = self._step_generator()

        while self.table[idx] is not None:
            if self.table[idx][0] == key:
                return self.table[idx]
            step = next(steps)
            idx = (idx + step) % self.size
            if idx == start_idx:
                return None
        return None

    def delete(self, key):
        idx = self._hash(key)
        start_idx = idx
        steps = self._step_generator()

        while self.table[idx] is not None:
            if self.table[idx][0] == key:
                self.table[idx] = None
                return
            step = next(steps)
            idx = (idx + step) % self.size
            if idx == start_idx:
                return
```

```python
# метод цепочек - Каждый индекс таблицы — это список, куда помещаются все пары с одинаковым остатком от деления ключа.
# При вставке, поиске и удалении происходит работа именно с этим списком, что обеспечивает корректное разрешение коллизий и позволяет хранить несколько элементов в одном бакете.
class HashTableChaining:
    def __init__(self, size=10):
        self.size = size
        self.table = [[] for _ in range(size)]

    def _hash(self, key):
        return key % self.size

    def insert(self, key, value=None):
        idx = self._hash(key)
        chain = self.table[idx]
        for i, (k, v) in enumerate(chain):
            if k == key:
                chain[i] = (key, value)
                return
        chain.append((key, value))

    def search(self, key):
        idx = self._hash(key)
        for (k, v) in self.table[idx]:
            if k == key:
                return (k, v)
        return None

    def delete(self, key):
        idx = self._hash(key)
        chain = self.table[idx]
        for i, (k, v) in enumerate(chain):
            if k == key:
                chain.pop(i)
                return
```

### Задание № 3:

<i>Расставить на стандартной 64-клеточной шахматной доске 8 ферзей так, чтобы ни один из них не находился под боем другого». Подразумевается, что ферзь бьёт все клетки, расположенные по вертикалям, горизонталям и обеим диагоналям
Написать программу,  которая находит хотя бы один способ решения задач.</i>



```python
def solve_eight_queens():
    n = 8
    board = [-1] * n  # Изначально ни в какой строке нет ферзя

    def is_safe(row, col):
        """
        Проверяет, можно ли безопасно поставить ферзя в позицию (row, col)
        без конфликтов с уже поставленными ферзями на предыдущих строках.
        """
        for i in range(row):
            # Проверяем, не находится ли ферзь в том же столбце
            # или на диагонали (разница позиций равна)
            if board[i] == col or abs(board[i] - col) == row - i:
                return False
        return True

    def backtrack(row):
        """
        Рекурсивно пытаемся расставить ферзей начиная с данной строки.
        Если ферзей расставлено во всех строках, возвращаем True.
        """
        if row == n:
            return True  # Найдено решение

        for col in range(n):
            if is_safe(row, col):
                board[row] = col  # Ставим ферзя в (row, col)
                if backtrack(row + 1):
                    return True
                board[row] = -1  # Убираем ферзя, откатываемся назад
        return False

    if backtrack(0):
        return board
    else:
        return None

# Получаем решение
solution = solve_eight_queens()

if solution:
    print("Найдено решение:")
    for i in range(8):
        row = ['.'] * 8
        row[solution[i]] = 'Q'
        print(' '.join(row))
else:
    print("Решение не найдено.")
```
